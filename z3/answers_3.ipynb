-- отрицание дизъюнкции есть конъюнкция отрицаний `p | q == ~ (~p & ~q)`


```python
p = Bool("p")
q = Bool("q")

prove( Or(p,q ) == Not(And(Not(p), Not(q))))
```


Найдите корень уравнения `x**3 + 3*x**2 + 4*x + 2 == 0` с помощью Z3. Сможете вы применить Z3 так, чтобы доказать, что найденный корень единственный, или опровергнуть это утверждение?

```python
x = Real("x")
solve(x**3 + 3*x**2 + 4*x + 2 == 0)

thm = Implies(x**3 + 3*x**2 + 4*x + 2 == 0, x==-1)

prove(thm)
```

находим контрпример

```python
thm = Implies(x**2 - 4 == 0, x==2)

prove(thm)
```

возьмем сумму углов треугольников как аксиому и докажем только один прямой угол

```python
A, B, C = Reals("A B C")
axioms = And(A + B + C == 180, A > 0, B > 0, C > 0)

# в треугольнике один прямой угол

thrm = Implies(And(axioms, A == 90), And(B < 90, C < 90))

prove(thrm)
```

если х четное, то х+2 четное

```python
x, y = Ints("x y")

prove(Implies(x%2==0, (x+2)%2==0))
```

доказательтво равенства условий после рефакторинга

```python
age = Int("age")

old  = And(age >= 18, age <= 65)
new = And(age > 17,  age < 66)

prove(old == new)
```
