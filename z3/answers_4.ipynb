разбор выражения по кусочкам

```python
x, y = Ints("x y")
e = x*2 + y

print(e.sort())
print(e.decl())
print(e.num_args())
print(e.arg(0), e.arg(1))
print(e.decl().kind() == Z3_OP_ADD)
```

считаем цену через substitute - подставляем числа вместо переменных

```python
x, y = Ints("x y")
price = x*100 + y*10

p = substitute(price, (x, IntVal(3)), (y, IntVal(5)))
print(p)
print(simplify(p))
```

если записать в массив два раза по одному индексу, то останется последнее значение

```python
A = Array("A", IntSort(), IntSort())
i, v1, v2 = Ints("i v1 v2")

prove(Store(Store(A, i, v1), i, v2)[i] == v2)
```

ищем массив из трех элементов - сумма 15, все по возрастанию, первый положительный

```python
X = IntVector("x", 3)

solve(Sum(X) == 15, X[0] < X[1], X[1] < X[2], X[0] > 0)
```

квадрат любого числа неотрицательный

```python
x = Real("x")

prove(ForAll(x, x*x >= 0))
```

если функция для всех аргументов возвращает 0, то и для 42 вернет 0

```python
f = Function("f", IntSort(), IntSort())
x = Int("x")

prove(Implies(ForAll(x, f(x) == 0), f(42) == 0))
```

минимальное колличество монет чтобы набрать 87 (монеты 1, 5, 10, 25)

```python
c1, c5, c10, c25 = Ints("c1 c5 c10 c25")

opt = Optimize()
opt.add(c1 >= 0, c5 >= 0, c10 >= 0, c25 >= 0)
opt.add(c1*1 + c5*5 + c10*10 + c25*25 == 87)
opt.minimize(c1 + c5 + c10 + c25)

print(opt.check())
print(opt.model())
```

грузим машину: большая коробка 20 кг и выгода 7, маленькая 5 кг и выгода 2. влезает 100 кг, маленьких на складе только 10. максимизируем выгоду

```python
big, small = Ints("big small")

opt = Optimize()
opt.add(big >= 0, small >= 0, small <= 10)
opt.add(big*20 + small*5 <= 100)
opt.maximize(big*7 + small*2)

print(opt.check())
print(opt.model())
```

копируем условия во второй солвер и добавляем жесткое условие - оригинал не ломается

```python
x = Int("x")

s1 = Solver()
s1.add(x > 0, x < 10)

s2 = Solver()
s2.add(s1.assertions())
s2.add(x > 20)

print(s1.check())
print(s2.check())
```

проверяем формулу двумя солверами: первый - бывает ли она истинной, второй - бывает ли ложной. второй выдал unsat, значит формула истинна всегда

```python
age = Int("age")
f = Implies(age >= 18, age > 10)

s1 = Solver()
s1.add(f)

s2 = Solver()
s2.add(Not(f))

print(s1.check())
print(s2.check())
```
